## Our Project
We decided to frame Expected Goals ($xG$) as a probabilistic binary classification task, evaluating model architectures on their ability to predict well-calibrated continuous shot probabilities using proper scoring rules, i.e. $\mathbb{P}(Y=1 | X)$.

xG is "Expected Goals", it measures the quality of a shot and its probability of resulting in a goal, instead of giving a shot a binary 0 - missed or 1 - goal it gives a situational probability, looking at the events just before the ball is struck. Comparing a team or player's actual goals relative to their xG can tell you how they are performing, if their $xG$ for a match was 2.0 but scored 3 they are over-performing, but if their $xG$ was 2 and they scored 1 they are under-performing.

We massaged our data, choosing what features we thought were relevant and neccessary for our models and which we could remove, ensuring we had no missing values or null cells, we also one-hot encoded any categorical feature data to make it binary, such as "shot_type".

We each chose a model to design: Luke chose random forests, Dylan chose a neural network, Don chose a linear regression and Pietro chose gradient boosting. Individually we analysed our models describing their flaws, comparing them to the gold standard xG model designed by the people from whomst we took the data. Then we had each model output it's predicted xG from which we calculated a Brier Score and a log-loss, in order to consistently evaluate them.

To ensure a fair test we all used the same cleaned data split the data into training, validating and testing in the same manner, so we all kept the 2025/26 season to test on.

To calculate the Brier Score we use the formula:
$$BS = \frac{1}{N}\sum_{t=1}^{N}{(f_t-\sigma_t)^2};$$
where $f_t$ is the predicted probability and $\sigma_t$ is the observed binary outcome, i.e. shot was a goal ($1$) or shot wasn't a goal ($0$), which yields an outcome between 0 and 1, where $BS\rightarrow0$ means the prediction is most accurate.


This scoring system penalises overconfidence as it calculates probabilistic distance between prediction and actual occurence, but is capped at a penalty of less than $1$, as an overconfident mistake; $f_t = 0.001$ and an outcome $\sigma_t = 1$ gives a penalty of $0.999^2$.

The log-loss idea is similar and is calculated by:
$$LogLoss = -\frac{1}{N}\sum_{i=1}^N\bigg{[}\sigma_tln({f_t})+(1-\sigma_t)ln(1-{f_t})\bigg{]}.$$
This measures the information-theoretic entropy, and assuming that a goal occurs ($\sigma_t = 1$) then the loss is $L = -\log{f_t}$. This means the loss is unbounded since the gradient $\frac{\partial L}{\partial f_t}=\frac{-1}{f_t}$, so as $f_t\rightarrow0^{+}$ the gradient approaches $+\infty$.

As a group we collectively analysed our models comparatively basing the Brier Scores and log-loss on the 2025/26 season.

We used these proper scoring rules as we do not want to categorise shots into binary goal or no-goal classes but instead as continuous conditional probabilities. Standard classification metrics like accuracy would not work due to class imbalance.

For example. across historical data for any given shot it has a $\approx10\%$ chance of resulting in a goal, if we evaluate this using classification accuracy with a standard threshold of $p\geq0.5$ where:
$$\text{Accuracy} = \frac{\text{True Positives}+\text{True Negatives}}{\text{Total Predictions}}$$
Then for a dummy model, that predicts $f_t=0$ for every shot, it has an accuracy of:
$$\frac{0+0.9N}{N}=90\%$$

Despite this model having a perceived 90% accuracy, it has evidently no predictive utility. Brier Score and log-loss solve this issue by heavily penalising naive baseline strategies, especially confident incorrect predictions, which giving every goal an $xG$ of 0, is.
